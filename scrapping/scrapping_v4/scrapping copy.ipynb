```python
import json

# Lire le fichier existant
with open('supply_chain_transparency_data.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Extraire seulement company et value
simplified_data = []
for item in data['items']:
    simplified_data.append({
        'company': item['company'],
        'value': item['value']
    })

# Sauvegarder le nouveau JSON simplifié
with open('transparency_scores_simple.json', 'w', encoding='utf-8') as f:
    json.dump(simplified_data, f, ensure_ascii=False, indent=2)

print(f"✅ {len(simplified_data)} entrées sauvegardées dans 'transparency_scores_simple.json'")

# Afficher un aperçu
print("\n📊 Aperçu des données simplifiées:")
for item in simplified_data[:5]:
    print(f"- {item['company']}: {item['value']}")
```
